# Data Selection with Pandas
## Data Mining Project — Student Performance Dataset

**Goal:** Select rows, columns, and subsets of data from a CSV file using Pandas.

> Before running the notebook, place your dataset CSV file in the same folder as this notebook. Update `FILE_PATH` below if your filename is different.


## 1. Import libraries
We use **Pandas** to load and select data, and **NumPy** for numerical operations.


```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)
```

## 2. Load the dataset
Change the file name below to match your CSV file exactly.


```python
FILE_PATH = "b119bb30-d943-4808-ab85-f269b231ebfb.csv"

df = pd.read_csv(FILE_PATH)

print("Dataset shape (rows, columns):", df.shape)
display(df.head())
```

## 3. Inspect column names
Run this cell first so you can see the exact column names in your dataset. Column names may differ between datasets.


```python
print("Columns:")
for column in df.columns:
    print(repr(column))

print("\\nData types:")
display(df.dtypes.to_frame("Data type"))
```

## 4. Select one column
Use single square brackets to select one column. The result is a Pandas Series.


```python
# Example: display the first column in the dataset
first_column = df.iloc[:, 0]
display(first_column.head())
print("Selected object type:", type(first_column))
```

## 5. Select multiple columns
Use a list of column names inside double square brackets. The code below selects the first two columns as a general example.


```python
selected_columns = df.iloc[:, :2]
display(selected_columns.head())
```

## 6. Select rows by position with `iloc`
`iloc` uses integer positions. For example, `df.iloc[0:5]` selects the first five rows.


```python
first_five_rows = df.iloc[0:5]
display(first_five_rows)
```

## 7. Select specific rows and columns with `iloc`
The example below selects the first five rows and first three columns.


```python
rows_and_columns = df.iloc[0:5, 0:3]
display(rows_and_columns)
```

## 8. Select columns by name with `loc`
Replace the example names with real column names from the dataset. The safe example below uses the first two column names automatically.


```python
chosen_names = df.columns[:2].tolist()
selected_by_name = df.loc[:, chosen_names]
print("Selected columns:", chosen_names)
display(selected_by_name.head())
```

## 9. Filter rows using a condition
Boolean filtering selects rows that satisfy a condition. The example chooses rows where the **first numeric column** is greater than its median, so it works even if column names vary.


```python
numeric_columns = df.select_dtypes(include="number").columns

if len(numeric_columns) > 0:
    numeric_col = numeric_columns[0]
    median_value = df[numeric_col].median()
    filtered_rows = df.loc[df[numeric_col] > median_value]
    print(f"Filter: {numeric_col} > median ({median_value:.2f})")
    display(filtered_rows.head(10))
else:
    print("No numeric columns were found. Choose a categorical column and filter it using an equality condition.")
```

## 10. Combine conditions
Use `&` for AND and `|` for OR. Put each condition in parentheses. This example selects rows above the median and below the maximum in the first numeric column.


```python
if len(numeric_columns) > 0:
    numeric_col = numeric_columns[0]
    median_value = df[numeric_col].median()
    max_value = df[numeric_col].max()
    combined = df.loc[
        (df[numeric_col] > median_value) &
        (df[numeric_col] < max_value)
    ]
    display(combined.head(10))
else:
    print("This example requires at least one numeric column.")
```

## 11. Select rows by category
This example uses the first non-numeric column and selects rows matching its most frequent non-missing value.


```python
categorical_columns = df.select_dtypes(exclude="number").columns

if len(categorical_columns) > 0:
    category_col = categorical_columns[0]
    common_value = df[category_col].mode(dropna=True)
    if not common_value.empty:
        value = common_value.iloc[0]
        category_rows = df.loc[df[category_col] == value]
        print(f"{category_col} == {value!r}")
        display(category_rows.head(10))
    else:
        print(f"{category_col} contains no non-missing values.")
else:
    print("No non-numeric columns were found.")
```

## 12. Select rows with `isin()`
`isin()` is useful when you want rows belonging to any value in a list. This example uses up to two of the most common values in the first categorical column.


```python
if len(categorical_columns) > 0:
    category_col = categorical_columns[0]
    top_values = df[category_col].value_counts(dropna=True).head(2).index.tolist()
    result = df.loc[df[category_col].isin(top_values)]
    print(f"Selected values from {category_col}:", top_values)
    display(result.head(10))
else:
    print("No non-numeric columns were found.")
```

## 13. Select rows by index labels with `loc`
`loc` uses index labels, while `iloc` uses integer positions. The default DataFrame index usually starts at 0.


```python
display(df.loc[df.index[:5]])
```

## 14. Create a new DataFrame from selected data
Keep only numeric columns as one example of preparing a subset for later analysis.


```python
numeric_df = df.select_dtypes(include="number").copy()
print("Numeric subset shape:", numeric_df.shape)
display(numeric_df.head())
```

## 15. Summary: common data-selection patterns

| Goal | Pandas example |
|---|---|
| One column | `df["column_name"]` |
| Multiple columns | `df[["column_a", "column_b"]]` |
| First five rows | `df.iloc[:5]` |
| Rows by position and columns | `df.iloc[0:5, 0:3]` |
| Columns by name | `df.loc[:, ["column_a", "column_b"]]` |
| Filter by condition | `df.loc[df["score"] >= 80]` |
| Combine conditions | `df.loc[(df["score"] >= 80) & (df["age"] > 18)]` |
| Match any listed value | `df.loc[df["group"].isin(["A", "B"])]` |

### Key point
Data selection extracts the rows and columns needed for a specific analysis without necessarily changing the original dataset.
